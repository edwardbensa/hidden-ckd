```python
import pandas as pd
```

```python
data = pd.read_csv("/Users/Edward/Downloads/open_postcode_geo.csv/open_postcode_geo.csv")
data
```

```python
df = pd.read_csv("/Users/Edward/Downloads/hiddenckd_postcodes.csv")
df.columns = ['date', 'event_address', 'event_postcode', 'gp_postcode']
df
```

```python
coord = data[['AB1 0AA', '57.101474', '-2.242851']].copy()
coord.columns = ['gp_postcode', 'lat', 'long']
```

```python
# Convert both to string (object) type
df['gp_postcode'] = df['gp_postcode'].astype(str)
coord['gp_postcode'] = coord['gp_postcode'].astype(str)

merged_df = df.join(coord.set_index('gp_postcode'), on='gp_postcode', how='left')
merged_df = merged_df.rename(columns={'lat':'gp_lat', 'long':'gp_long'})
coord = coord.rename(columns={'gp_postcode':'event_postcode'})
merged_df = merged_df.join(coord.set_index('event_postcode'), on='event_postcode', how='left')
merged_df = merged_df.rename(columns={'lat':'event_lat', 'long':'event_long'})
merged_df
```

```python
import numpy as np

# Radius of Earth in kilometers
R = 6371.0

merged_df['event_lat'] = pd.to_numeric(merged_df['event_lat'], errors='coerce')
merged_df['event_long'] = pd.to_numeric(merged_df['event_long'], errors='coerce')
merged_df['gp_lat'] = pd.to_numeric(merged_df['gp_lat'], errors='coerce')
merged_df['gp_long'] = pd.to_numeric(merged_df['gp_long'], errors='coerce')

# Convert degrees to radians
event_lat_rad = np.radians(merged_df['event_lat'])
event_long_rad = np.radians(merged_df['event_long'])
gp_lat_rad = np.radians(merged_df['gp_lat'])
gp_long_rad = np.radians(merged_df['gp_long'])

# Differences
delta_lat = gp_lat_rad - event_lat_rad
delta_long = gp_long_rad - event_long_rad

# Haversine formula
a = np.sin(delta_lat / 2)**2 + np.cos(event_lat_rad) * np.cos(gp_lat_rad) * np.sin(delta_long / 2)**2
c = 2 * np.arcsin(np.sqrt(a))

# Distance in kilometers
merged_df['distance_km'] = R * c
```

```python
merged_df = merged_df.drop(columns=['gp_lat', 'gp_long', 'event_lat', 'event_long'])
```

```python
merged_df.to_csv('/Users/Edward/Downloads/hiddenckd_travel_distances.csv', index=False)
```
